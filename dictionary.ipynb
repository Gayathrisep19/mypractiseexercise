```python
#Creating empty dict
student={}
student={'Name':'Gayathri','Age':33,'City':'Chennai','Language':['English','Tamil']}
student
#length of dictionary
print(len(student))
#view keys of dictionary
print(student.keys())
#view values of dictionary
print(student.values())
#view key value pairs in dictionary ,gives key value as tuples
print(student.items())
```

```python
#add new key to dicitionary
student['pet']='dog'
print(student)
#to update existing key value
student['pet']=['dog','cat']
print(student)
#delete key from dicitionary
del student['pet']
print(student)
#to update using update method
student.update({'pet':'dog'})
print(student)
#remove a key from dicitionary and also to save it in variable
pet=student.pop('pet')
print(student)
print(pet)
#Retrieve a key value from dicitionary
print(student['Name'])
#Retrieve a key value from dicitionary without getting error if key not found(default to None)
print(student.get('pet'))
#Retrieve a key value from dicitionary with default value if key not found
print(student.get('pet','dog'))
```

```python
#Iterate through keys
for key in student.keys():
    print(key,student[key])
#Iterate through values
for value in student.values():
    print(value)
#Iterte through pair
for key,value in student.items():
    print(key,value)
#Sort dict based on key alphabhetically
sortedstudent={key:val for key,val in sorted(student.items())}
print(sortedstudent)
population={'India':1836183513,'China':79777967,'US':87916680}
#Sort dict based on value ascending
sortedpopulation={key:val for key,val in sorted(population.items(),key=lambda val:val[1])}
print(sortedpopulation)
#Sort dict based on value descending
sortedpopulation={key:val for key,val in sorted(population.items(),key=lambda val:val[1],reverse=True)}
print(sortedpopulation)
```
