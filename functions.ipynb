```python
def returndouble(n):
    return n*2
def hello(msg):
    print(msg)
def argumentfunction(a,b):
    c=a+b
    return c
def firstdigit(num):
    fdigit=str(num)[0]
    return fdigit
def repeatstr(n):
    return "GFG" * n
def printupto(n):
    for i in range(0,n+1):
        print(i,end=" ")
    print("\n")
def fibonaciseries(n):
    i=0
    firstnum,secondnum=0,1
    print(firstnum,end=" ")
    print(secondnum,end=" ")
    while i < n:
        nextnum=firstnum+secondnum
        print(nextnum,end=" ")
        firstnum=secondnum
        secondnum=nextnum
        i=i+1
    
```

```python
res=returndouble(3)
print(res)
hello("Hi am gayathri")
print(argumentfunction(4,5))
print(firstdigit(9789))
print(repeatstr(3))
printupto(5)
fibonaciseries(5)
```
