```python
colors=['Red','Blue','Green']
#Length of the list
print(len(colors))
#print entire list
print(colors)
#Access list based on index
print(colors[1])
print(colors[0])
#Access list using negative index
print(colors[-1])
print(colors[-2])
print(colors[-3])
```

```python
#List slicing,start is inclusive and end index is exclusive
print(colors[0:1])
#this includes complete list
print(colors[:])
#This includes from first index till end
print(colors[1:])

#Add element to list
colors.append('Yellow')
#Add element at particular position
colors.insert(1,'purple')
#Add multiple elements to list
colors.extend(['Orange','Black'])
#Append adds another list if it is multiple element to add
colors.append(['White','Indigo'])
print(colors)
```

```python
#Remove element from list 
colors.pop(-2)

#remove multiple elements from list
#del colors[-1:]
colors
```

```python
#Get the index of the searched element
colors.index('Red')

#Check if element in list
print('Green' in colors)

#Loop through list
for element in colors:
    print(element)
    
#Loop through with iteration number:
for num,element in enumerate(colors):
    print(f"We have {element} at {num} position")
```

```python
Numericals=[19,53,1,28,926,45]
#Sort the elements in list inplace ascending
Numericals.sort(reverse=False)
Numericals
#Sort the elements in list inplace descending
Numericals.sort(reverse=True)
Numericals

#To not modify existing list & return sorted value as new list
Numericals_sorted=sorted(Numericals)
Numericals_sorted

#To replace prticular index in list
Numericals[1]=10
Numericals
```
